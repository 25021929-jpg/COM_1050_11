```python
# Nhập danh sách các số nguyên từ bàn phím
nums = list(map(int, input().split()))

result = []
seen = set() # Dùng set để kiểm tra sự tồn tại cho nhanh (O(1))

for num in nums:
    if num not in seen:
        result.append(num)
        seen.add(num)

print(result)
```

```python
# Dòng 1: Nhập tuple
data = tuple(map(int, input().split()))
# Dòng 2: Nhập k
k = int(input())

# Xử lý trường hợp danh sách rỗng để tránh lỗi
if len(data) > 0:
    # Nếu k lớn hơn độ dài tuple, ta lấy phần dư để tối ưu vòng xoay
    k = k % len(data)

    # Thực hiện cắt (slicing) và nối
    # data[k:] là phần sau, data[:k] là phần đầu bị đẩy ra sau
    result = data[k:] + data[:k]
    print(result)
else:
    print(data)
```

```python
# Nhập chuỗi các cặp key:value
raw_input = input().split()

dictionary = {}

for pair in raw_input:
    # Tách key và value bởi dấu hai chấm
    key, value = pair.split(':')

    # Nếu key chưa có trong từ điển, tạo list rỗng
    if key not in dictionary:
        dictionary[key] = []

    # Thêm value vào list của key đó
    dictionary[key].append(value)

print(dictionary)
```

```python
# Nhập danh sách các chuỗi
words = input().split()

# Khởi tạo dictionary rỗng
word_count = {}

for word in words:
    # Nếu từ đã có trong từ điển thì cộng thêm 1
    if word in word_count:
        word_count[word] += 1
    # Nếu chưa có thì gán bằng 1
    else:
        word_count[word] = 1

print(word_count)
```

```python
# Nhập danh sách các số từ bàn phím
nums = list(map(int, input().split()))

evens = [] # Danh sách chứa số chẵn
odds = []  # Danh sách chứa số lẻ

for num in nums:
    if num % 2 == 0:
        evens.append(num)
    else:
        odds.append(num)

# Chuyển đổi list về tuple và in ra
print(tuple(evens))
print(tuple(odds))
```

```python
# Nhập danh sách số
nums = list(map(int, input().split()))

# Bước 1: Đếm tần suất
counts = {}
for num in nums:
    if num in counts:
        counts[num] += 1
    else:
        counts[num] = 1

# Nếu danh sách rỗng thì dừng (để tránh lỗi)
if not counts:
    print(None)
else:
    # Bước 2: Tìm tần suất lớn nhất
    max_freq = max(counts.values())

    # Bước 3: Tìm tất cả các số có tần suất bằng max_freq
    candidates = []
    for num, freq in counts.items():
        if freq == max_freq:
            candidates.append(num)

    # Bước 4: In ra số nhỏ nhất trong các ứng viên
    print(min(candidates))
```

```python
# Nhập chuỗi dữ liệu thô (dạng: key1 value1 key2 value2 ...)
raw_data = input().split()

inverted_dict = {}

# Duyệt qua danh sách với bước nhảy là 2 (để lấy từng cặp key-value)
# i là vị trí của key, i+1 là vị trí của value
for i in range(0, len(raw_data), 2):
    original_key = raw_data[i]
    original_value = raw_data[i+1]

    # Thực hiện đảo ngược: Value cũ thành Key mới, Key cũ thành Value mới
    inverted_dict[original_value] = original_key

print(inverted_dict)
```

```python
# Nhập hai dòng dữ liệu
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))

# Chuyển list2 thành set để tìm kiếm nhanh hơn (O(1))
set2 = set(list2)
result = []

for num in list1:
    # Nếu num có trong list2 và chưa có trong kết quả
    if num in set2 and num not in result:
        result.append(num)

print(result)
```

```python
# Giả sử dòng đầu nhập n (số hàng) và m (số cột)
n, m = map(int, input().split())

matrix = []
# Nhập n hàng
for _ in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

# In ra ma trận
for row in matrix:
    # Duyệt từng phần tử trong hàng
    for item in row:
        # {:>4} nghĩa là dành 4 chỗ trống và căn lề phải
        print(f"{item:>4}", end="")
    print() # Xuống dòng sau khi in xong 1 hàng
```

```python
#In ra đường chéo chính và đường chéo phụ của mảng\
n,m = map(int,input().split())
matrix = []
for _ in range(n):
  row = list(map(int,input().split()))
  matrix.append(row)
for row in range(n):
  for col in range(m):
    if col == row: #Đường chéo chính
      print(matrix[row][col],end = " ")
    elif row + col == m-1: #đường chéo phụ
      print(matrix[row][col],end = " ")
    else:
      print(" ",end = " ")
  print()
```

```python
# Nhập n (hàng), m (cột), k (chỉ số cột cần tính)
n, m, k = map(int, input().split())

matrix = []
for _ in range(n):
    matrix.append(list(map(int, input().split())))

column_sum = 0

# Kiểm tra k có hợp lệ không (0 <= k < m)
if 0 <= k < m:
    for i in range(n):
        column_sum += matrix[i][k] # Cộng dồn phần tử tại cột k của hàng i
    print(column_sum)
else:
    print("Chỉ số cột k không hợp lệ")
```

```python
# Nhập chuỗi các từ cách nhau bởi dấu cách
words = input().split()

# Khởi tạo dictionary kết quả
result = {}

# Dùng enumerate để duyệt cả vị trí (index) và từ (word)
for index, word in enumerate(words):
    # Nếu từ chưa có trong dict, tạo list rỗng
    if word not in result:
        result[word] = []

    # Thêm vị trí hiện tại vào list của từ đó
    result[word].append(index)

# Chuyển đổi list thành tuple theo yêu cầu đề bài
for key in result:
    result[key] = tuple(result[key])

print(result)
```
